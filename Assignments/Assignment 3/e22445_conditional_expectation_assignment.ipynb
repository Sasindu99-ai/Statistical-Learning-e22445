# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

#### Answer for Q2

To derive the conditional distribution $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, we start with the joint density function of $\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix}$:
$$f_{\mathbf{Z}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

We can partition the mean and covariance matrix as follows:
$$\boldsymbol{\mu} = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix}, \quad \boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}.$$
Thus,
$$\boldsymbol{\mu} = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) \\ \boldsymbol{\mu}_Y(\mathbf{y}) \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) \end{bmatrix}.$$
Thus,
$$\boldsymbol{\mu} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) + \boldsymbol{\Sigma}_{YY}(\mathbf{y} - \boldsymbol{\mu}_y) \end{bmatrix}.$$
Thus, $$\boldsymbol{\mu} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) + \boldsymbol{\Sigma}_{YY}(\mathbf{y} - \boldsymbol{\mu}_y) \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) + \boldsymbol{\Sigma}_{YY}(\mathbf{y} - \boldsymbol{\mu}_y) \end{bmatrix} = \begin{bmatrix} \boldsymbol{\mu}_X(\mathbf{y}) + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y) \\ \boldsymbol{\mu}_Y(\mathbf{y}) + \boldsymbol{\Sigma}_{YY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y) \end{bmatrix}.$$

Next, we can express the joint density function in terms of the conditional density and the marginal density:
$$f_{\mathbf{Z}}(\mathbf{z}) = f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) f_{\mathbf{Y}}(\mathbf{y}).$$
The marginal density of $\mathbf{Y}$ is given by:
$$f_{\mathbf{Y}}(\mathbf{y}) = \frac{1}{\sqrt{(2\pi)^m |\boldsymbol{\Sigma}_{YY}|}} \exp\left( -\frac{1}{2} (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right).$$
Thus, we can express the conditional density as:
$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{Z}}(\mathbf{z})}{f_{\mathbf{Y}}(\mathbf{y})}.$$
Substituting the expressions for $f_{\mathbf{Z}}(\mathbf{z})$ and $f_{\mathbf{Y}}(\mathbf{y})$, we get:
$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T (\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX})^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
$$\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX},$$
and
$$\boldsymbol{\mu}^* = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y).$$
Thus, we have shown that the conditional distribution of $\mathbf{X}$ given $\mathbf{Y}$ is a multivariate normal distribution with mean $\boldsymbol{\mu}^*$ and covariance matrix $\boldsymbol{\Sigma}_{X\mid Y}$.

Summarizing, the conditional density is given by:
$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
$$\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX},$$
and
$$\boldsymbol{\mu}^* = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y).$$

# Q3) Continuous Fault Monitoring of a Spindle Motor

 A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**
1.  **Hidden Temperature ($X$):** The standardized temperature deviation follows a standard normal distribution:
    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

---

**Answer the followings:**

**1. Joint Distribution**
Identify the distribution of the random varaible $(X, Y)$.
* Show that $Y \sim \mathscr{N}(0, 5)$.
* Find the covariance $\text{Cov}(X, Y)$.

**2. Information Geometry**
Explain why $X$ is **not** $\sigma(Y)$-measurable. In practical terms, if the power meter shows $Y = 5$, why can't the engineer know for certain if the motor is overheating ($X$ is high) or if the material being cut is simply very hard ($Q$ is high)?

**3. Conditional Expectation (The Best Estimate)**
Compute the conditional expectation $E[X \mid Y]$.
* Using the property for joint normals: $E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$.
* Express this as a "signal-to-total-variance" ratio.

**4. Verification of the Residual**
Let $\widehat{X} = E[X \mid Y]$. Define the estimation error (residual) as $Z = X - \widehat{X}$.
* Prove that $E[Z] = 0$.
* Verify that the error $Z$ is uncorrelated with the observation $Y$ (the Orthogonality Principle).

**5. Information Theory: Differential Entropy**
* Compute the differential entropy $h(X)$.
* Compute the conditional differential entropy $h(X \mid Y)$.
* *Note: The variance of the conditional distribution $(X \mid Y=y)$ is given by $\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)}$.*

**6. Mutual Information**
Calculate $I(X; Y) = h(X) - h(X \mid Y)$.
* What percentage of the "uncertainty" about the temperature is removed by observing the power meter?
* How would this change if the mechanical noise $Q$ had a variance of $100$ instead of $4$?

**7. Diagnostic Interpretation**
The factory manager says: *"Since $Y = X + Q$, our best estimate of temperature should just be $Y$ itself."*
Using your result from Question 3, explain why $Y$ is a **biased** estimator of $X$ and why $E[X \mid Y]$ "shrinks" the measurement toward the mean (zero).



#### Answers for Q3

1. **Joint Distribution**
- To find the distribution of $Y$, we note that $Y = X + Q$, where $X \sim \mathscr{N}(0, 1)$ and $Q \sim \mathscr{N}(0, 4)$ are independent. The sum of two independent normal random variables is also normally distributed, with mean equal to the sum of the means and variance equal to the sum of the variances. Thus:
$$Y \sim \mathscr{N}(0 + 0, 1 + 4) = \mathscr{N}(0, 5).$$
- To find the covariance $\text{Cov}(X, Y)$, we use the definition of covariance:
$$\text{Cov}(X, Y) = E[XY] - E[X]E[Y].$$
Since $E[X] = 0$ and $E[Y] = 0$, we have:
$$\text{Cov}(X, Y) = E[X(X + Q)] = E[X^2] + E[XQ].$$
Since $X$ and $Q$ are independent, $E[XQ] = E[X]E[Q] = 0$. Thus:
$$\text{Cov}(X, Y) = E[X^2] = \text{Var}(X) = 1.$$

2. **Information Geometry**
$X$ is not $\sigma(Y)$-measurable because $Y$ is a noisy observation of $X$. The power meter reading $Y$ is influenced by both the internal temperature $X$ and the mechanical load $Q$. If the power meter shows $Y = 5$, the engineer cannot determine whether this is due to a high temperature (overheating) or a high mechanical load (hard material) because both $X$ and $Q$ contribute to $Y$. The noise from $Q$ obscures the true value of $X$, making it impossible to infer $X$ with certainty from $Y$ alone.

3. **Conditional Expectation (The Best Estimate)**
Using the property for joint normals, we have:
$$E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y).$$
Substituting the values:
$$E[X \mid Y = y] = 0 + \frac{1}{5}(y - 0) = \frac{1}{5}y.$$
This can be expressed as a "signal-to-total-variance" ratio:
$$E[X \mid Y] = \frac{\text{Var}(X)}{\text{Var}(X) + \text{Var}(Q)} Y = \frac{1}{1 + 4} Y = \frac{1}{5} Y.$$

4. **Verification of the Residual**
- To prove that $E[Z] = 0$, we calculate:
$$E[Z] = E[X - \widehat{X}] = E[X] - E[E[X \mid Y]] = E[X] - E[X] = 0.$$
- To verify that $Z$ is uncorrelated with $Y$, we calculate:
$$\text{Cov}(Z, Y) = E[ZY] - E[Z]E[Y] = E[(X - \widehat{X})Y] - 0 \cdot 0 = E[XY] - E[\widehat{X}Y].$$
Since $\widehat{X} = E[X \mid Y]$, we have $E[\widehat{X}Y] = E[E[X \mid Y]Y] = E[X Y]$. Thus:
$$\text{Cov}(Z, Y) = E[XY] - E[XY] = 0.$$
This confirms that $Z$ is uncorrelated with $Y$.

5. **Information Theory: Differential Entropy**
- The differential entropy of a normal distribution $X \sim \mathscr{N}(\mu, \sigma^2)$ is given by:
$$h(X) = \frac{1}{2} \log(2\pi e \sigma^2).$$
For $X \sim \mathscr{N}(0, 1)$, we have:
$$h(X) = \frac{1}{2} \log(2\pi e \cdot 1) = \frac{1}{2} \log(2\pi e) \approx 1.4189 \text{ nats}.$$
- The variance of the conditional distribution $(X \mid Y=y)$ is given by:
$$\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)} = 1 - \frac{1^2}{5} = 1 - \frac{1}{5} = \frac{4}{5}.$$
Thus, the conditional differential entropy is:
$$h(X \mid Y) = \frac{1}{2} \log(2\pi e \sigma_{X|Y}^2) = \frac{1}{2} \log\left(2\pi e \cdot \frac{4}{5}\right) = \frac{1}{2} \log\left(\frac{8\pi e}{5}\right) \approx 1.2925 \text{ nats}.$$

6. **Mutual Information**
The mutual information is calculated as:
$$I(X; Y) = h(X) - h(X \mid Y) = 1.4189 - 1.2925 = 0.1264 \text{ nats}.$$
To find the percentage of uncertainty about the temperature removed by observing the power meter, we calculate:
$$\text{Percentage} = \frac{I(X; Y)}{h(X)} \times 100\% = \frac{0.1264}{1.4189} \times 100\% \approx 8.91\%.$$
If the mechanical noise $Q$ had a variance of $100$ instead of $4$, the variance of $Y$ would be $1 + 100 = 101$, and the covariance would still be $1$. Thus, the conditional variance would be:
$$\sigma_{X|Y}^2 = 1 - \frac{1^2}{101} = 1 - \frac{1}{101} = \frac{100}{101}.$$
The conditional differential entropy would then be:
$$h(X \mid Y) = \frac{1}{2} \log\left(2\pi e \cdot \frac{100}{101}\right) \approx 1.4189 \text{ nats}.$$
Thus, the mutual information would be:
$$I(X; Y) = h(X) - h(X \mid Y) = 1.4189 - 1.4189 = 0 \text{ nats}.$$
This means that if the mechanical noise is very high, observing the power meter provides no information about the temperature, and the percentage of uncertainty removed would be 0%.
