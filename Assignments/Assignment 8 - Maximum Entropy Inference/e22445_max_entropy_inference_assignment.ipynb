# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

### 1. The Partition Function

The canonical partition function is the sum over all states:

$$Z(T) = \sum_{j=0}^\infty e^{-\frac{E_j}{k_B T}}$$

Let $\beta = \frac{1}{k_B T}$. Substituting the energy eigenvalues $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$:

$$Z = \sum_{j=0}^\infty e^{-\beta \hbar \omega \left(j + \frac{1}{2}\right)}$$

We can factor out the constant term (the zero-point energy contribution) that does not depend on the summation index $j$:

$$Z = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^\infty \left(e^{-\beta \hbar \omega}\right)^j$$

The summation is a standard geometric series of the form $\sum_{j=0}^\infty r^j = \frac{1}{1-r}$, where the common ratio is $r = e^{-\beta \hbar \omega}$. Since $\beta \hbar \omega > 0$, we know $r < 1$, meaning the series converges:

$$Z = e^{-\frac{\beta \hbar \omega}{2}} \left( \frac{1}{1 - e^{-\beta \hbar \omega}} \right)$$

Substituting $\beta = \frac{1}{k_B T}$ back into the equation yields the target expression:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

### 2. Thermodynamic Potentials

**Helmholtz Free Energy ($F$)**

Using the relation $F(T) = -k_B T \ln Z$:

$$F = -k_B T \ln\left( \frac{e^{-\frac{\beta \hbar \omega}{2}}}{1 - e^{-\beta \hbar \omega}} \right)$$

Using logarithmic properties ($\ln(a/b) = \ln(a) - \ln(b)$):

$$F = -k_B T \left[ -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right]$$

Substitute $\beta = \frac{1}{k_B T}$ back in:

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

**Internal Energy ($U$)**

Using the relation $U(T) = -\frac{\partial \ln Z}{\partial \beta}$:

$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln(1 - e^{-\beta \hbar \omega})$$

Differentiating with respect to $\beta$:

$$U = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \frac{\partial}{\partial \beta} \left(1 - e^{-\beta \hbar \omega}\right) \right]$$

$$U = \frac{\hbar \omega}{2} + \frac{1}{1 - e^{-\beta \hbar \omega}} \left(\hbar \omega e^{-\beta \hbar \omega}\right)$$

Divide the numerator and denominator of the second term by $e^{-\beta \hbar \omega}$ to simplify:

$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$



### 3. Entropy

Using the relation $S = -\left(\frac{\partial F}{\partial T}\right)$ and applying the chain rule to our expression for $F(T)$:

$$S = -\frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

Apply the product rule to the second term:

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \left[ \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \left(-e^{-\frac{\hbar \omega}{k_B T}}\right) \left(\frac{\hbar \omega}{k_B T^2}\right) \right]$$

Simplify the second term (noting that the two negative signs cancel):

$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{1}{T} \frac{\hbar \omega e^{-\frac{\hbar \omega}{k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

Divide the top and bottom of the fraction by $e^{-\frac{\hbar \omega}{k_B T}}$:

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{1}{T} \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

**Verification ($U = F + TS$):**

Let's calculate $F + TS$:

$$F + TS = \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{1}{T} \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right]$$

The terms with the natural logarithm cancel out perfectly:

$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

This matches our derivation for $U(T)$ exactly.



### 4. Heat Capacity and Hyperbolic Form

First, calculate $C_V = \left(\frac{\partial U}{\partial T}\right)$. We differentiate the internal energy (the zero-point energy term drops out as a constant):

$$C_V = \frac{\partial}{\partial T} \left[ \hbar \omega \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^{-1} \right]$$

$$C_V = \hbar \omega \left(-1\right) \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^{-2} e^{\frac{\hbar \omega}{k_B T}} \left(-\frac{\hbar \omega}{k_B T^2}\right)$$

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

To rewrite this using hyperbolic functions, we manipulate the fraction. Let $x = \frac{\hbar \omega}{2k_B T}$:

$$\frac{e^{2x}}{(e^{2x} - 1)^2} = \frac{e^{2x}}{\left(e^x (e^x - e^{-x})\right)^2} = \frac{e^{2x}}{e^{2x} (e^x - e^{-x})^2} = \frac{1}{(e^x - e^{-x})^2}$$

Using the identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we know that $e^x - e^{-x} = 2\sinh(x)$. Therefore:

$$\frac{1}{(e^x - e^{-x})^2} = \frac{1}{(2\sinh(x))^2} = \frac{1}{4\sinh^2(x)}$$

Substitute this back into our $C_V$ expression, where $4$ can be absorbed into the squared term:

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{4\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$


### 5. Asymptotic Behavior (Discussion)

In the high-temperature limit, thermal energy dominates the quantum energy spacing, meaning $k_B T \gg \hbar \omega$.

Let's define a small parameter $x = \frac{\hbar \omega}{2k_B T}$. In the high-temperature limit, $x \ll 1$.

The Taylor series expansion for the hyperbolic sine for small arguments is $\sinh(x) \approx x$. Substituting this into our hyperbolic $C_V$ formula yields:

$$C_V \approx k_B (x)^2 \frac{1}{(x)^2} = k_B$$

**Does it match the classical prediction?**

Yes. According to the classical equipartition theorem, each quadratic degree of freedom contributes $\frac{1}{2}k_B T$ to the internal energy. A 1D harmonic oscillator has two quadratic degrees of freedom: kinetic energy ($\propto p^2$) and potential energy ($\propto x^2$).

Therefore, classical internal energy is $U_{classical} = \frac{1}{2}k_B T + \frac{1}{2}k_B T = k_B T$. Differentiating with respect to temperature gives $C_V = k_B$. The quantum heat capacity perfectly asymptotes to the classical limit when the temperature is high enough to wash out the discrete nature of the energy levels.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

To find the probability density function $f(x)$ that maximizes the differential entropy $H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) dx$ subject to the normalization constraint $\int_{-\infty}^{\infty} f(x) dx = 1$, the mean constraint $\int_{-\infty}^{\infty} (x - \mu) f(x) dx = 0$, and the variance constraint $\int_{-\infty}^{\infty} (x - \mu)^2 f(x) dx = \sigma^2$, we use Lagrange multipliers.

The Lagrangian functional is:
$$\mathcal{L}(f) = \int_{-\infty}^{\infty} \left[ -f(x) \ln f(x) + \lambda_0 f(x) + \lambda_1 (x - \mu) f(x) + \lambda_2 (x - \mu)^2 f(x) \right] dx$$

Setting the functional derivative to zero:
$$\frac{\delta \mathcal{L}}{\delta f} = -\ln f(x) - 1 + \lambda_0 + \lambda_1(x - \mu) + \lambda_2(x - \mu)^2 = 0$$

Solving for $f(x)$:
$$f(x) = \exp\left( (\lambda_0 - 1) + \lambda_1(x - \mu) + \lambda_2(x - \mu)^2 \right)$$

For $f(x)$ to be a valid, normalizable distribution that satisfies the mean constraint, it must be symmetric around $\mu$, requiring $\lambda_1 = 0$. Also, $\lambda_2$ must be negative to prevent divergence. Letting $C = \exp(\lambda_0 - 1)$, we have:
$$f(x) = C \exp\left( \lambda_2(x - \mu)^2 \right)$$

Applying the normalization constraint $\int_{-\infty}^{\infty} C \exp\left( \lambda_2(x - \mu)^2 \right) dx = 1$ gives:
$$C \sqrt{-\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{-\frac{\lambda_2}{\pi}}$$

Applying the variance constraint $\int_{-\infty}^{\infty} (x - \mu)^2 C \exp\left( \lambda_2(x - \mu)^2 \right) dx = \sigma^2$ yields:
$$C \left( -\frac{1}{2\lambda_2} \sqrt{-\frac{\pi}{\lambda_2}} \right) = \sigma^2$$

Substituting $C$:
$$\sqrt{-\frac{\lambda_2}{\pi}} \left( -\frac{1}{2\lambda_2} \sqrt{-\frac{\pi}{\lambda_2}} \right) = \sigma^2 \implies -\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = -\frac{1}{2\sigma^2}$$

Substituting $\lambda_2$ back into $C$:
$$C = \sqrt{-\frac{1}{\pi} \left(-\frac{1}{2\sigma^2}\right)} = \frac{1}{\sqrt{2\pi\sigma^2}}$$

Therefore, substituting $C$ and $\lambda_2$ into $f(x)$ gives the maximum entropy distribution:
$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




The AI feature extractor produces a vector $\mathbf{x} = (x_1, x_2, x_3)^T$. According to the Maximum Entropy principle for constrained expected values, the probability of the email being routed to Technical Support ($Y=1$) takes the form of a Logistic Regression (sigmoid) model:

$$P(Y=1|\mathbf{x}) = \sigma(\mathbf{w}^T \mathbf{x}) = \frac{1}{1 + \exp(- (w_1 x_1 + w_2 x_2 + w_3 x_3))}$$

Let $p_s = P(Y=1|\mathbf{x}_s)$ represent the conditional probability for each operational scenario. The theoretical probabilities evaluate to:
* $p_A = \sigma(w_1)$
* $p_B = \sigma(w_2)$
* $p_C = \sigma(w_3)$
* $p_D = \sigma(w_2 + w_3)$
* $p_E = \sigma(w_1 + w_2)$

Using the Law of Total Expectation, the historical constraints $\mathbb{E}[x_j \cdot \mathbb{I}(Y=1)] = c_j$ establish the following non-linear system of equations:

**Constraint 1 ($x_1$ - Financial):**
$$P(\mathbf{x}_A)p_A + P(\mathbf{x}_E)p_E = 0.05 \implies 0.40 \cdot \sigma(w_1) + 0.05 \cdot \sigma(w_1 + w_2) = 0.05$$

**Constraint 2 ($x_2$ - Account Access):**
$$P(\mathbf{x}_B)p_B + P(\mathbf{x}_D)p_D + P(\mathbf{x}_E)p_E = 0.15 \implies 0.25 \cdot \sigma(w_2) + 0.15 \cdot \sigma(w_2 + w_3) + 0.05 \cdot \sigma(w_1 + w_2) = 0.15$$

**Constraint 3 ($x_3$ - Bug/Malfunction):**
$$P(\mathbf{x}_C)p_C + P(\mathbf{x}_D)p_D = 0.16 \implies 0.05 \cdot \sigma(w_3) + 0.15 \cdot \sigma(w_2 + w_3) = 0.16$$

Because these are non-linear, we solve for the optimal weights $\mathbf{w} = (w_1, w_2, w_3)$ numerically.

```python
import numpy as np
from scipy.optimize import root

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def equations(w):
    w1, w2, w3 = w

    # Probabilities for each scenario based on current weights
    p_A = sigmoid(w1)
    p_B = sigmoid(w2)
    p_C = sigmoid(w3)
    p_D = sigmoid(w2 + w3)
    p_E = sigmoid(w1 + w2)

    # The 3 historical expectation constraints
    eq1 = 0.40 * p_A + 0.05 * p_E - 0.05
    eq2 = 0.25 * p_B + 0.15 * p_D + 0.05 * p_E - 0.15
    eq3 = 0.05 * p_C + 0.15 * p_D - 0.16

    return [eq1, eq2, eq3]

# Solve the system of non-linear equations using an initial guess of 0s
initial_w = [0.0, 0.0, 0.0]
solution = root(equations, initial_w)

if solution.success:
    w1, w2, w3 = solution.x

    print("=== Optimal Logistic Weights ===")
    print(f"w1 (Financial Context) : {w1:7.4f}")
    print(f"w2 (Access Context)    : {w2:7.4f}")
    print(f"w3 (Bug Context)       : {w3:7.4f}\n")

    print("=== Final Routing Probabilities to Technical Support (Y=1) ===")
    print(f"Scenario A (Pure Financial)   [1,0,0] : {sigmoid(w1):.2%}")
    print(f"Scenario B (Pure Access)      [0,1,0] : {sigmoid(w2):.2%}")
    print(f"Scenario C (Pure Technical)   [0,0,1] : {sigmoid(w3):.2%}")
    print(f"Scenario D (Access Technical) [0,1,1] : {sigmoid(w2 + w3):.2%}")
    print(f"Scenario E (Billing Access)   [1,1,0] : {sigmoid(w1 + w2):.2%}")

    # Verify the constraints
    print("\n=== Verifying Constraints ===")
    c1 = 0.40 * sigmoid(w1) + 0.05 * sigmoid(w1 + w2)
    c2 = 0.25 * sigmoid(w2) + 0.15 * sigmoid(w2 + w3) + 0.05 * sigmoid(w1 + w2)
    c3 = 0.05 * sigmoid(w3) + 0.15 * sigmoid(w2 + w3)
    print(f"Constraint 1 (Target = 0.05) : {c1:.4f}")
    print(f"Constraint 2 (Target = 0.15) : {c2:.4f}")
    print(f"Constraint 3 (Target = 0.16) : {c3:.4f}")
else:
    print("Solver failed to converge.")
```

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



### Maximum Entropy Distribution for Asset Price $S_T$

To find the continuous probability density function $p(S_T)$ that maximizes the differential entropy subject to the given market constraints, we use the method of Lagrange multipliers.

**1. The Objective Function and Constraints**

The differential entropy to be maximized for $S_T \in (0, \infty)$ is:
$$H(p) = -\int_0^\infty p(S_T) \ln p(S_T) dS_T$$

Subject to four constraints:
*   **Normalization:** $\int_0^\infty p(S_T) dS_T = 1$
*   **Mean:** $\int_0^\infty S_T p(S_T) dS_T = 102$
*   **Second Moment:** $\int_0^\infty S_T^2 p(S_T) dS_T = 10600$
*   **Log Moment (Skew):** $\int_0^\infty \ln(S_T) p(S_T) dS_T = 4.60$

**2. Setting up the Lagrangian**

We define the Lagrangian functional $\mathcal{L}$ with multipliers $\lambda_0$, $\lambda_1$, $\lambda_2$, and $\lambda_3$:
$$\mathcal{L}(p) = \int_0^\infty \left[ -p(S_T) \ln p(S_T) + \lambda_0 p(S_T) + \lambda_1 S_T p(S_T) + \lambda_2 S_T^2 p(S_T) + \lambda_3 \ln(S_T) p(S_T) \right] dS_T$$

**3. Finding the Extremum**

To find the function $p(S_T)$ that maximizes this integral, we take the functional derivative with respect to $p(S_T)$ and set it to zero:
$$\frac{\delta \mathcal{L}}{\delta p} = -\ln p(S_T) - 1 + \lambda_0 + \lambda_1 S_T + \lambda_2 S_T^2 + \lambda_3 \ln(S_T) = 0$$

Solving for $p(S_T)$:
$$\ln p(S_T) = (\lambda_0 - 1) + \lambda_1 S_T + \lambda_2 S_T^2 + \lambda_3 \ln(S_T)$$
$$p(S_T) = \exp(\lambda_0 - 1) \exp\left( \lambda_1 S_T + \lambda_2 S_T^2 + \lambda_3 \ln(S_T) \right)$$

**4. Final Functional Form**

Let $Z = \exp(1 - \lambda_0)$ act as the normalization constant (the partition function). Using the logarithmic property $\exp(\lambda_3 \ln(S_T)) = S_T^{\lambda_3}$, the maximum entropy distribution simplifies to:

$$p(S_T) = \frac{1}{Z} S_T^{\lambda_3} \exp\left(\lambda_1 S_T + \lambda_2 S_T^2\right)$$

Where the partition function is defined as:
$$Z(\lambda_1, \lambda_2, \lambda_3) = \int_0^\infty S_T^{\lambda_3} \exp\left(\lambda_1 S_T + \lambda_2 S_T^2\right) dS_T$$

**Mathematical Limitations and Numerical Solution:**
Unlike the standard Gaussian normal distribution, the addition of the logarithmic constraint creates a state that lacks a closed-form algebraic solution for the multipliers.

*   To ensure the distribution does not diverge to infinity as $S_T \to \infty$, we require $\lambda_2 < 0$.
*   To ensure the distribution is integrable as $S_T \to 0$, we require $\lambda_3 > -1$.

The exact numerical values for the Lagrange multipliers $\mathbf{\lambda} = (\lambda_1, \lambda_2, \lambda_3)$ are found by minimizing the dual convex optimization problem, enforcing the system of partial derivatives:
$$\frac{\partial \ln Z}{\partial \lambda_1} = 102$$
$$\frac{\partial \ln Z}{\partial \lambda_2} = 10600$$
$$\frac{\partial \ln Z}{\partial \lambda_3} = 4.60$$
