````python
class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }
````

## Question 1. 2.

### Why does Sensor_4’s noise inflate PCA eigenvalues?

---

### 1. Key structural fact: PCA optimizes total variance

PCA solves the eigen-decomposition of the sample covariance matrix:

$$
\mathbf{S} = \frac{1}{n-1} X^T X
$$

and finds directions $\mathbf{v}$ maximizing:

$$
\max_{\|\mathbf{v}\|=1} \mathbf{v}^T \mathbf{S} \mathbf{v}
$$

So PCA has **no mechanism to distinguish**:

- structural (latent) variance
- noise (idiosyncratic variance)

It only sees:

> “variance = signal”

---

### 2. Effect of Sensor_4: variance domination

Sensor 4 is defined as:

$$
X_4 = 0.02 f_1 + 0.05 f_2 + \epsilon_4
$$

with:

$$
\mathrm{Var}(\epsilon_4) \gg \mathrm{Var}(f_1), \mathrm{Var}(f_2)
$$

So:

$$
\mathrm{Var}(X_4) \approx \mathrm{Var}(\epsilon_4)
$$

This creates a **high-energy axis in raw covariance space**.

---

### 3. Geometric effect on covariance structure

The sample covariance matrix becomes:

$$
\mathbf{S} =
\underbrace{\mathbf{S}_{\text{structure}}}_{\text{low-rank signal}}
+
\underbrace{\mathbf{S}_{\text{noise}}}_{\text{diagonal-dominant distortion}}
$$

Sensor 4 contributes a large diagonal term:

$$
S_{44} \gg S_{11}, S_{22}, S_{33}
$$

So the covariance ellipsoid becomes:

- stretched strongly along Sensor_4 axis
- rotated toward noise direction

---

### 4. Why PCA is “tricked”

PCA interprets variance magnitude as importance.

So the algorithm:

1. Sees inflated variance in Sensor_4 direction
2. Rotates PC1 toward maximizing projection onto that noisy axis
3. Redistributes remaining structure into subsequent PCs

Result:

- **PC1 captures noise-dominated direction**
- **PC2–PC3 become contaminated mixtures of structure + noise**

---

### 5. Effect on eigenvalues (key point)

Eigenvalues satisfy:

$$
\lambda_1 \ge \lambda_2 \ge \cdots \ge \lambda_m
$$

Because Sensor_4 inflates total variance:

- $\lambda_1$ increases artificially
- $\lambda_2$ also increases due to rotation of mixed variance
- true latent structure is no longer cleanly separated

So:

> PCA eigenvalues are “inflated” because noise adds high-energy directions that PCA treats as meaningful structure.

---

### 6. Engineering interpretation (critical insight)

This is not a mathematical failure — it is a **modeling mismatch**:

- PCA assumes:
  $$
  \text{All variance is informative}
  $$

- But your system violates this:
  $$
  \text{Variance} = \text{structure} + \text{fault noise}
  $$

So Sensor_4 acts as a:

> **variance amplifier that distorts the global covariance geometry**

---

### Final Answer (concise)

Sensor 4 inflates PCA eigenvalues because its high-variance noise dominates the covariance matrix, forcing PCA to align its principal components with this artificial high-energy direction. Since PCA maximizes total variance without separating signal from noise, it incorrectly interprets Sensor 4’s stochastic variance as a dominant structural mode, thereby distorting and inflating the leading eigenvalues and contaminating the learned subspace.

Explain how the **Varimax rotation** in Factor Analysis relaxes this rule to create "simple structure" (clean, isolated groupings).

## Question 2. 1.

### How Varimax rotation creates “simple structure” by relaxing PCA’s variance hierarchy

---

### 1. PCA constraint: variance maximization with rigid orthogonality

PCA constructs eigenvectors of the covariance matrix:

$$
\mathbf{S} \mathbf{v}_j = \lambda_j \mathbf{v}_j
$$

with enforced ordering:

$$
\lambda_1 > \lambda_2 > \cdots > \lambda_m
$$

Each principal component is chosen to:

$$
\max_{\mathbf{v}_j} \mathrm{Var}(\mathbf{v}_j^T X)
$$

subject to orthogonality constraints.

### Key consequence:
- Each PC is a **global variance aggregator**
- Loadings are typically **dense (mixed contributions from all sensors)**

So PCA prioritizes:

> variance concentration over interpretability

---

### 2. Factor Analysis shift: decoupling “variance” from “interpretability”

Factor Analysis starts from:

$$
\mathbf{R} = \Lambda \Lambda^T + \Psi
$$

Unlike PCA, the factor solution is **not unique**.

Any orthogonal transformation $T$ gives:

$$
\Lambda^* = \Lambda T, \quad \text{where } T^T T = I
$$

and still preserves:

$$
\Lambda \Lambda^T = \Lambda^* (\Lambda^*)^T
$$

So FA introduces a **rotation freedom** that PCA does not exploit.

---

### 3. Varimax objective: enforce sparse loadings

Varimax chooses $T$ to maximize:

$$
\sum_{j=1}^{m} \left[
\frac{1}{m} \sum_{r=1}^{k} \lambda_{jr}^{*4}
- \left(\frac{1}{m} \sum_{r=1}^{k} \lambda_{jr}^{*2}\right)^2
\right]
$$

### Intuition of this objective:

It maximizes variance of squared loadings across factors, which leads to:

- large loadings become larger
- small loadings shrink toward zero

---

### 4. What this changes geometrically

Instead of PCA’s:

- “spread variance evenly across all sensors”

Varimax produces:

- each sensor loads strongly on **one or few factors**
- weak cross-loadings are suppressed

So the loading matrix becomes:

$$
\Lambda_{\text{Varimax}} \approx
\begin{bmatrix}
\text{strong} & 0 \\
\text{strong} & 0 \\
0 & \text{strong} \\
0 & \text{noise}
\end{bmatrix}
$$

---

### 5. Relaxation of PCA’s hierarchy

PCA enforces:

- global ordering by $\lambda_j$
- components are ranked by total variance

Varimax instead:

- **does not reorder by eigenvalue magnitude**
- redistributes variance across rotated axes
- prioritizes **interpretability over maximal variance capture**

So the rule becomes:

> “Do not maximize variance per component — maximize clarity of variable-to-factor assignment.”

---

### Final Answer (concise)

Varimax rotation relaxes PCA’s strict eigenvalue hierarchy by allowing orthogonal rotations of the loading matrix that preserve total variance but redistribute it across factors. Instead of maximizing variance concentration in ordered principal components, Varimax optimizes a sparsity criterion that forces loadings to become highly polarized-driving each sensor to load strongly on a single factor and weakly on others. This produces a “simple structure” where latent factors correspond to clean, physically interpretable groupings rather than mixed variance directions.

## Question 2. 2.

### Why is the rotated FA heatmap easier to troubleshoot than a raw PCA eigenvector matrix?

---

### 1. PCA eigenvectors are *mathematically optimal but physically entangled*

A PCA loading matrix is constructed as:

$$
\mathbf{P} = [\mathbf{v}_1, \mathbf{v}_2, \dots, \mathbf{v}_m]
$$

where each $\mathbf{v}_j$ maximizes variance:

$$
\mathbf{v}_j = \arg\max_{\|\mathbf{v}\|=1} \mathbf{v}^T \mathbf{S} \mathbf{v}
$$

### Operational consequence:

Each principal component becomes a **mixture of all sensors**:

$$
PC_j = a_{j1}X_1 + a_{j2}X_2 + a_{j3}X_3 + a_{j4}X_4
$$

So even if a failure originates in one sensor, its effect is:

- distributed across multiple PCs
- partially hidden inside global variance structure

---

### 2. Factor Analysis rotation isolates physical groupings

After Varimax rotation:

$$
\Lambda^* = \Lambda T
$$

This preserves covariance structure:

$$
\Lambda \Lambda^T = \Lambda^* (\Lambda^*)^T
$$

but changes *representation*.

---

### 3. Result: sparse “sensor-to-factor” mapping

The rotated FA heatmap tends to produce:

- Sensor_1 → strongly Factor_1
- Sensor_2 → strongly Factor_1
- Sensor_3 → strongly Factor_2
- Sensor_4 → weak / noise-only factor contributions

So the heatmap becomes:

- block-structured
- sparse
- directly interpretable

---

### 4. Why this matters during failure diagnosis

From a plant operator’s perspective, failure diagnosis requires:

#### (a) Localization
Identify *which sensor / subsystem is failing*

FA provides:

- direct mapping: sensor → latent physical mode
- clear separation of structural vs noise channels

PCA provides:

- distributed influence across multiple PCs
- ambiguous attribution of failure source

---

#### (b) Fault interpretability

If Sensor_4 degrades:

- FA: it appears as **high uniqueness ($\varphi^2$)** immediately
- PCA: it contaminates multiple eigenvectors, making root cause unclear

---

#### (c) Reduced cognitive load

FA heatmap is:

- sparse
- clustered
- approximately block-diagonal in structure

PCA matrix is:

- dense
- globally mixed
- requires post-hoc interpretation

---

### 5. Geometric interpretation

- PCA rotates axes to maximize variance spread → **global mixing**
- FA (Varimax) rotates axes to maximize sparsity → **local grouping**

So FA aligns axes closer to:

> physical subsystems of the plant

rather than:

> statistical variance directions

---

### Final Answer (concise)

The rotated Factor Analysis heatmap is easier to troubleshoot because Varimax rotation produces a sparse, block-structured loading matrix that maps sensors cleanly onto latent physical factors. This allows direct identification of which subsystem is responsible for a structural failure. In contrast, PCA eigenvectors are dense linear mixtures optimized for variance, causing fault signatures to spread across multiple components and making root-cause diagnosis ambiguous and operationally inefficient.

## Question 4.1.

### PCA Strategy: $T^2$ + $Q$ Monitoring

---

### 1. What PCA monitoring is doing

PCA-based monitoring splits behavior into two orthogonal metrics:

#### In-subspace variation (Hotelling’s $T^2$):

$$
T^2_i = \sum_{j=1}^{k} \frac{z_{i,j}^2}{\lambda_j}
$$

Measures:
- deviation **within learned normal subspace**
- changes in dominant system dynamics

---

#### Out-of-subspace variation (SPE / $Q$):

$$
Q_i = \sum_{j=k+1}^{m} z_{i,j}^2
$$

Measures:
- reconstruction error
- energy not explained by PCA model

---

### 2. Operational interpretation

- $T^2$ → structured shifts in operating conditions
- $Q$ → emergence of new/unmodelled behavior

---

### Conclusion

The PCA strategy monitors system health by separating behavior into two parts: $T^2$ tracks how much the system deviates within the learned normal operating space, while $Q$ measures how much behavior cannot be explained by that space. Together, they provide a geometric view of deviations from normality, where $T^2$ captures internal shifts and $Q$ captures external or novel anomalies.

## Question 4.2.

### FA Strategy: Monitoring Latent Factor Scores

---

### 1. What FA monitoring uses

Factor Analysis operates on latent scores:

$$
\mathbf{F} = \mathbf{Z}\Lambda
$$

where:
- $\mathbf{Z}$ = standardized observations
- $\Lambda$ = rotated loading matrix

Each sample is projected into **latent physical factors**.

---

### 2. Monitoring principle

Instead of splitting variance like PCA, FA monitors:

#### Latent factor stability:

$$
F_{i,r} \quad \text{for } r = 1, \dots, k
$$

and typically tracks:

- factor score shifts
- factor variance drift
- cross-factor decoupling

---

### 3. Operational meaning

Each factor corresponds to a **physical subsystem mode**, e.g.:

- structural load mode ($f_1$)
- operational dynamics ($f_2$)

So monitoring becomes:

> “Are the underlying physical drivers changing?”

---

### Conclusion

The FA strategy monitors system health by tracking the evolution of latent factor scores, which represent underlying physical processes driving the sensors. Instead of decomposing deviations into geometric variance components like PCA, FA directly observes shifts in these latent physical modes, making it more interpretable and better aligned with real system behavior and fault sources.

FA is more robust because $\varphi^2$ acts like a “noise container” for each sensor, preventing a single faulty channel from corrupting the shared latent structure. PCA lacks this separation, so it allows one bad sensor to influence the entire model.
